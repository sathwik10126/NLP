```python
medical_text = """John Doe, a 52-year-old male, was admitted to City Hospital on 15/08/2024 with complaints of chest pain and shortness of breath.
                 His blood pressure was 150/95 mmHg, and he has a history of diabetes mellitus and hypertension.
                 The doctor prescribed metformin 500 mg twice daily and advised a follow-up after two weeks."""
```

**Download packages**

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
words=word_tokenize(medical_text)
```

### **Tokenization**

```python
from nltk.tokenize import sent_tokenize
sent_tokenize(medical_text)
```

```python
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
words_in_quote = word_tokenize(medical_text)
words_in_quote
```

# **Filtering StopWords**

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words = set(stopwords.words("english"))
filtered_list = []
for word in words_in_quote:
  if word.casefold() not in stop_words:
    filtered_list.append(word)
filtered_list
```

### **Stemming**

## **Porter Stemmer**

```python

from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
stemmer = PorterStemmer()
words = word_tokenize(medical_text)
stemmed_words = [stemmer.stem(word) for word in words]
stemmed_words
```

## **Snowball Stemmer**

```python
from nltk.stem import SnowballStemmer
from nltk.tokenize import word_tokenize
stemmer = SnowballStemmer('english')
words = word_tokenize(medical_text)
stemmed_words = [stemmer.stem(word) for word in words]
stemmed_words
```

# **Lancaster Stemmer**

```python
from nltk.stem import LancasterStemmer
from nltk.tokenize import word_tokenize
stemmer = LancasterStemmer()
words = word_tokenize(medical_text)
stemmed_words = [stemmer.stem(word) for word in words]
stemmed_words
```

### **Lemmatization**

```python
nltk.download('omw-1.4')
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
words = word_tokenize(medical_text)
for word in words:
    print(word,"--->",lemmatizer.lemmatize(word))
```

### **Comparison**

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer
from nltk.tokenize import word_tokenize

porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer(language='english')
words_from_medical_text = [word.lower() for word in word_tokenize(medical_text) if word.isalpha()]

print("{0:20}{1:20}{2:20}{3:30}".format("Word","Porter Stemmer","Snowball Stemmer","Lancaster Stemmer"))
print("-"*130)
for word in words_from_medical_text:
    print("{0:20}{1:20}{2:20}{3:30}".format(word,porter.stem(word),snowball.stem(word),lancaster.stem(word)))
```
