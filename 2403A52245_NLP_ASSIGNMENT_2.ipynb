```python
medical_text = """John Doe, a 52-year-old male, was admitted to City Hospital on 15/08/2024 with complaints of chest pain and shortness of breath.
                 His blood pressure was 150/95 mmHg, and he has a history of diabetes mellitus and hypertension.
                 The doctor prescribed metformin 500 mg twice daily and advised a follow-up after two weeks."""
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
words=word_tokenize(medical_text)
```

```python
from nltk.tokenize import sent_tokenize
sent_tokenize(medical_text)
```

```python
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
words_in_quote = word_tokenize(medical_text)
words_in_quote
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words = set(stopwords.words("english"))
filtered_list = []
for word in words_in_quote:
  if word.casefold() not in stop_words:
    filtered_list.append(word)
filtered_list
```

```python

from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
stemmer = PorterStemmer()
words = word_tokenize(medical_text)
stemmed_words = [stemmer.stem(word) for word in words]
stemmed_words
```

```python
nltk.download('omw-1.4')
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
words = word_tokenize(medical_text)
for word in words:
    print(word,"--->",lemmatizer.lemmatize(word))
```
